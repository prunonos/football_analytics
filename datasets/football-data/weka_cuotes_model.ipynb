```python
def getPoints(lista, local):
    if local:
        l = 0
    else:
        l = 1
        
    lista[lista == 'D'] = 1
    lista[lista == 'H'] = 3 + (l * -3)
    lista[lista == 'A'] = 0 + (l *  3)
    
    print(lista)
    
    try:
            avg = sum(lista)/len(lista)
    except ZeroDivisionError:
            avg = 0.0
            pass
    
    print(avg)
    
    return avg
```

```python
def calculoRendimiento(data,n=5):
    
    rend_local = []
    rend_visit = []
    encajados_h = []
    marcados_h = []
    encajados_a = []
    marcados_a = []
    shots_h = []
    shots_a = []

    for index in range(len(data)):
        local = data.loc[index]['HomeTeam']
        visitante = data.loc[index]['AwayTeam']

    #     LOCAL:
        indices = []
        indices = [i for i, x in enumerate(data['HomeTeam']) if x == local and i > index and len(indices) < 5]
        print(local + ' - local')
        indices = indices[:n]
        
#       Puntos
        pts = data['FTR'][indices]
        avg_pts = getPoints(pts,local=True)
        rend_local.append(avg_pts)
        
#       Goles Marcados y Recibidos
        marc = data['FTHG'][indices]
        marc = sum(marc)
        print(marc)
        marcados_h.append(marc)
        recib = data['FTAG'][indices]
        recib = sum(recib)
        encajados_h.append(recib)
        
        # Disparos
        shots = data['HS'][indices]
        
        try:
            shots = sum(shots)/len(shots)
        except ZeroDivisionError:
            shots = 0.0
            pass
        
        print(shots)
        shots_h.append(shots)

    #     VISITANTE:
        indices = []
        indices = [i for i, x in enumerate(data['AwayTeam']) if x == visitante and i > index]
        print(visitante + ' - visitante')
        indices = indices[:n]

        res = data['FTR'][indices]
        avg = getPoints(res,local=False)
        rend_visit.append(avg)
#       Goles Marcados y Recibidos
        marc = data['FTAG'][indices]
        marc = sum(marc)
        marcados_a.append(marc)
        recib = data['FTHG'][indices]
        recib = sum(recib)
        encajados_a.append(recib)

        # Disparos
        shots = data['AS'][indices]
        try:
            shots = sum(shots)/len(shots)
        except ZeroDivisionError:
            shots = 0.0
            pass        
        shots_a.append(shots)
    
    return [rend_local,rend_visit,marcados_h,encajados_h,marcados_a,encajados_a,shots_h,shots_a]
```
