```python
from collections import defaultdict
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains

# Ruta al controlador del navegador (especifica la ruta correcta)
driver_path = 'C:\\Users\\GuillemUPV\\chromedriver-win64\\chromedriver.exe'
# Inicializar el navegador
driver = webdriver.Chrome()

# URL de la página web
url = 'https://www.whoscored.com/Matches/738817/Live/Spain-LaLiga-2013-2014-Malaga-Levante'
# Acceder a la página web
driver.get(url)

# Esperar a que la página cargue completamente (puedes ajustar este tiempo según sea necesario)
driver.implicitly_wait(2)

players = defaultdict(lambda: defaultdict(dict))

for tipo in ['.pitch-field', '.bench']:
    # Encontrar todos los elementos de jugador o banquillo
    equipo_H, equipo_A = driver.find_elements(By.CSS_SELECTOR, tipo)

    equipo_id_H = equipo_H.get_attribute("data-team-id")
    equipo_id_A = equipo_A.get_attribute("data-team-id")
    field_H = equipo_H.get_attribute('data-field')
    field_A = equipo_A.get_attribute('data-field')
    
    jugadores_H = equipo_H.find_elements(By.CSS_SELECTOR, f'.player[data-field="{field_H}"]')
    jugadores_A = equipo_A.find_elements(By.CSS_SELECTOR, f'.player[data-field="{field_A}"]')

    # Iterar sobre los jugadores titulares e imprimir sus datos
    for jugador_H,jugador_A in zip(jugadores_H,jugadores_A):

        ActionChains(driver).click(jugador_H).perform()
        ActionChains(driver).click(jugador_A).perform()
        
        # Esperar a que aparezca la ventana emergente (puedes ajustar este tiempo según sea necesario)
        driver.implicitly_wait(2)

        for jugador in [jugador_H,jugador_A]:
            player_id = jugador.get_attribute("data-player-id")
            field = jugador.get_attribute("data-field")

            # Obtener datos de la ventana emergente
            player_info = driver.find_element(By.CSS_SELECTOR, f'.compared[data-mode="player"][data-field={field}]')
            position = player_info.find_element(By.CSS_SELECTOR, '.player-position').text
            age = player_info.find_element(By.CSS_SELECTOR, '.player-age').text
            height = player_info.find_element(By.CSS_SELECTOR, '.player-height').text
            weight = player_info.find_element(By.CSS_SELECTOR, '.player-weight').text

            # Cerrar la ventana emergente
            # driver.find_element(By.CSS_SELECTOR, '.close-comparison').click()

            # Obtener otros datos del jugador
            nombre = jugador.find_element(By.CSS_SELECTOR, '.player-name').text
            numero_camiseta = jugador.find_element(By.CSS_SELECTOR, '.shirt-number').text
            puntaje = jugador.find_element(By.CSS_SELECTOR, '.player-stat-value').text

            minutes_played = 90 if tipo == '.pitch-field' else 0

            # Agregar datos a la estructura
            players[player_id]["team_id"] = equipo_id
            players[player_id]["field"] = field
            players[player_id]["player_id"] = player_id
            players[player_id]["nombre"] = nombre
            players[player_id]["titular"] = int(tipo == '.pitch-field')
            players[player_id]["rating"] = puntaje
            players[player_id]["minutes_played"] = minutes_played
            players[player_id]["position"] = position
            players[player_id]["age"] = age
            players[player_id]["height"] = height
            players[player_id]["weight"] = weight

            # Obtener datos de estadísticas
            stats_container = driver.find_element(By.CSS_SELECTOR, '.match-centre-stats[data-mode="player"]')
            stats = stats_container.find_elements(By.CSS_SELECTOR, 'li[class="match-centre-stat match-centre-sub-stat"]')

            for stat in stats:
                stat_name = stat.get_attribute('data-for')
                stat_value = stat.find_element(By.CSS_SELECTOR,f'.match-centre-stat-value[data-field="{field}"]').get_attribute('data-value')
                players[player_id][stat_name] = stat_value
                print(field,nombre,stat_name,stat_value)

# Cerrar el navegador
driver.quit()

df_players = pd.DataFrame(players).T
```
