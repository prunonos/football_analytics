```python
import pandas as pd
import os
from datetime import datetime
```

```python
path_data = "F:\\TFG\datasets\\football-data\\"
path_raw = "F:\\TFG\\datasets\\raw_datasets\\"
```

```python
df_concat = pd.DataFrame([])
for s in os.listdir(path_data):
    if '.csv' in s:
        try:
            df = pd.read_csv(path_data + s, sep=',')[['Div','Date','HomeTeam','AwayTeam','FTR','B365D','B365H','B365A']]
            df_concat = pd.concat([df_concat,df],axis=0).reset_index(drop=True)
        except:
            print(s)
            # continue
```

```python
df_concat.Date.loc[0]
```

```python
def str2date(d):
    if isinstance(d,float): return None
    try:
        return datetime.strptime(d,"%d/%m/%y")
    except:
        return datetime.strptime(d,"%d/%m/%Y")
```

```python
df_concat['Date'] = df_concat.Date.apply(str2date)
df_concat = df_concat.dropna()
```

```python
df_concat['res'] = df_concat.FTR.map({'H':1,'D':0,'A':2})
```

```python
path_train      = 'F:\\TFG\\datasets\\raw_datasets\\'
raw_Data  = pd.read_csv(path_train+'historical_goals_date'+'.csv',sep=';',decimal=',',index_col='matchId')#.set_index('wyId')


cols = ['FTR',
'FTHG',
'FTAG',
'aux'
]

raw_Data = raw_Data.drop(cols,axis=1,errors='ignore')
raw_Data.sort_index()

raw_Data['year'] = raw_Data.Date.map(lambda x: (datetime.strptime(x,"%Y-%m-%d")).year)
```

```python
raw_Data = raw_Data[['Date','HomeTeam','AwayTeam']].reset_index()
raw_Data['Date'] = raw_Data.Date.apply(lambda d: datetime.strptime(d,"%Y-%m-%d"))
```

```python
raw_Data
```

```python
df_concat
```

```python
df_concat = df_concat.merge(raw_Data, on=['Date','HomeTeam','AwayTeam'])
df_concat
```

```python
df_concat['draw_pred'] = 1/df_concat.B365D
df_concat['home_pred'] = 1/df_concat.B365H
df_concat['away_pred'] = 1/df_concat.B365A
df_concat['label'] = df_concat.res

df_concat
```

```python
df_concat['prediction'] = df_concat[['draw_pred','home_pred','away_pred']].to_numpy().argmax(axis=1)
```

```python
df_concat[['matchId','draw_pred','home_pred','away_pred','prediction','label']].to_csv(path_raw + 'bets.csv',sep=';',decimal=',')
```
