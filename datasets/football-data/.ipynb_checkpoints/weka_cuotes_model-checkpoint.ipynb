```python
ej = dataframe1.loc[:379]
ej = ej.sort_index(ascending=False,ignore_index=True)
ej
```

```python
# rendimiento = np.zeros((380,2),int)
rend_local = []
rend_visit = []
np = 5

for index in range(len(ej)):
    local = ej.loc[index]['HomeTeam']
    visitante = ej.loc[index]['AwayTeam']
    
#     LOCAL:
    indices = [i for i, x in enumerate(ej['HomeTeam']) if x == local and i > index]
    print(local + ' - local')
    indices = indices[:np]
    
    res = ej['FTR'][indices]
    res = toPoints(res,local=True)
    try:
        avg = sum(res)/len(res)
    except ZeroDivisionError:
        avg = 0.0
        pass
    rend_local.append(avg)
    print(avg)
    
    
#     VISITANTE:
    indices = [i for i, x in enumerate(ej['AwayTeam']) if x == visitante and i > index]
    print(visitante + ' - visitante')
    indices = indices[:np]
    
    res = ej['FTR'][indices]
    res = toPoints(res,local=False)
    try:
        avg = sum(res)/len(res)
    except ZeroDivisionError:
        avg = 0.0
        pass
    rend_visit.append(avg)
    print(avg)
```

```python
assert len(rend_local) == len(rend_visit) == len(ej) == 380
```

```python
def toPoints(lista, local):
    if local:
        l = 0
    else:
        l = 1
        
    lista[lista == 'D'] = 1
    lista[lista == 'H'] = 3 + (l * -3)
    lista[lista == 'A'] = 0 + (l *  3)
    
    return lista
```

```python
#LALIGA
for temp, num in zip(lista_temporadas,range(0,25)):
    for file in os.listdir(os.getcwd()):
        aux = '(' + str(num) + ')'
        if file.find('SP1 ' + aux) == False:
            os.rename('SP1 ' + aux + '.csv','SP1_' + temp + '.csv')

ligas['LaLiga'] = {}

for temp in lista_temporadas:
    try:
        ligas['LaLiga'][temp] = pd.read_csv(os.getcwd()+'\\SP1_'+ temp +'.csv')
        ligas['LaLiga'][temp] = ligas['LaLiga'][temp].sort_index(ascending=False,ignore_index=True)
    except Exception:
        print(temp + ' da error')
        continue
```

```python
def calculoRendimientoPuntos(data,n=5):
    rend_local = []
    rend_visit = []

    for index in range(len(data)):
        local = data.loc[index]['HomeTeam']
        visitante = data.loc[index]['AwayTeam']

    #     LOCAL:
        indices = [i for i, x in enumerate(data['HomeTeam']) if x == local and i > index]
#         print(local + ' - local')
        indices = indices[:n]

        res = data['FTR'][indices]
        res = toPoints(res,local=True)
        try:
            avg = sum(res)/len(res)
        except ZeroDivisionError:
            avg = 0.0
            pass
        rend_local.append(avg)
#         print(avg)


    #     VISITANTE:
        indices = [i for i, x in enumerate(data['AwayTeam']) if x == visitante and i > index]
#         print(visitante + ' - visitante')
        indices = indices[:n]

        res = data['FTR'][indices]
        res = toPoints(res,local=False)
        try:
            avg = sum(res)/len(res)
        except ZeroDivisionError:
            avg = 0.0
            pass
        rend_visit.append(avg)
#         print(avg)
    
    return [rend_local,rend_visit]
```

```python
lista_ligas = list(ligas['LaLiga'].values())
data = pd.concat(lista_ligas[:10], axis=0, ignore_index=True)
data
rend = calculoRendimientoPuntos(data)
```
