```python
import pandas as pd

path_footdata   ='F:\\TFG\\datasets\\football-data\\cache\\'


def correct_csv(input_path, output_path, read=True):
    with open(input_path, 'r', encoding='ISO-8859-1') as infile:
        lines = infile.readlines()

    # Determine the correct number of columns based on the header
    header = lines[0]
    num_columns = len(header.split(','))

    corrected_lines = [header]

    for line in lines[1:]:
        columns = line.split(',')
        if len(columns) != num_columns:
            # Try to correct the line by joining extra columns
            corrected_line = ','.join(columns[:num_columns])
            corrected_lines.append(corrected_line + '\n')  # Add newline character
        else:
            corrected_lines.append(line)
    
    # Write the corrected lines to a new file
    with open(output_path, 'w', encoding='ISO-8859-1') as outfile:
        for line in corrected_lines:
            outfile.write(line)
    
    print(f"Corrected file saved to {output_path}")

    if read:
        return pd.read_csv(output_path,sep=',',on_bad_lines='error',encoding='latin-1')


# Input and output file paths
input_path = f'{path_footdata}E1_0203.csv'
output_path = f'{path_footdata}corrected\\E1_0203_corrected.csv'

# Correct the CSV file
correct_csv(input_path, output_path)

# Read the corrected CSV file with pandas
# df = pd.read_csv(output_path, encoding='ISO-8859-1')
# print(df.head())
```
