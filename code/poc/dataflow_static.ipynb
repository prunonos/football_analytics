```python
columns_to_keep = [
    'Div','Date','Attendance','HomeTeam','AwayTeam','FTHG','FTAG','FTR','HTHG','HTAG','HTR','Referee',
    'HS','AS','HST','AST','HF','AF','HC','AC','HY','AY','HR','AR','HO','AO','HHW','AHW','B365H','B365D','B365A', 'MaxH','MaxD','MaxA','AvgH','AvgD','AvgA'
]
```

```python
import os
data = pd.DataFrame(data=[],columns=columns_to_keep)

for elem in os.listdir(path_footdata):
    if (".csv" in elem):
        # print(elem)
        df = pd.read_csv(path_footdata+elem,sep=',',on_bad_lines='skip',encoding='latin-1')
        df = df.loc[:, df.columns.isin(columns_to_keep)]

        season = f"T{elem[-8:-6]}-{elem[-6:-4]}"
        df["season"] = season
        print(df.Div[0],season,df.shape)
        
        data = pd.concat([data,df],axis=0,ignore_index=True)
```

```python
data
data['Date'] = pd.to_datetime(data['Date'],dayfirst=True)
data = data[data.Date.isna()==False]
data
```

```python
data['div_order'] = data.groupby(['Div']).cumcount()
data.groupby(['Div']).div_order.max()
data
```

```python
dict_teams = {
    "M'Gladbach":"M'gladbach",
    "Extremadura UD":"Extremadura",
    "Orense":"Ourense",
    "Villareal":"Villarreal"
}

data["HomeTeam"] = data["HomeTeam"].apply(lambda t: dict_teams.get(t,t))
data["AwayTeam"] = data["AwayTeam"].apply(lambda t: dict_teams.get(t,t))
data
```

```python
teams_id = data[['Div','HomeTeam']].drop_duplicates().sort_values(['Div','HomeTeam'])
teams_id["AwayTeam"] = teams_id.HomeTeam
teams_id['id'] = np.arange(len(teams_id))
# teams_id = teams_id.set_index(["Div","HomeTeam"])
# teams_id = teams_id.to_dict()['id']
teams_id
```

```python
teams_id[teams_id.id==288]
# teams_id[teams_id.HomeTeam=="Extremadura"]
```

```python
cols = ['Div','HomeTeam']
data["id_H"] = data.set_index(cols).merge(teams_id[[*cols,'id']],on=cols,how='left').id.to_list()
cols = ['Div','AwayTeam']
data["id_A"] = data.set_index(cols).merge(teams_id[[*cols,'id']],on=cols,how='left').id.to_list()

data
```

```python
cols_metadata = ['matchId','id_H', 'id_A','Div','div_order','Date','season','HomeTeam', 'AwayTeam', 'FTHG', 'FTAG', 'FTR','Attendance','Referee']
cols_features = ['HS', 'AS', 'HST', 'AST', 'HF','AF', 'HC', 'AC', 'HY', 'AY', 'HR', 'AR','HO','AO','HHW','AHW']
cols_bets     = ['B365H', 'B365D', 'B365A','MaxH', 'MaxD', 'MaxA', 'AvgH', 'AvgD', 'AvgA']
```

```python
data = data[[*cols_metadata,*cols_features,*cols_bets]] 
```
