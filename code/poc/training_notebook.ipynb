```python
import training
import pandas as pd
import torch
import optuna
import utils as ut
import dataflow_own as own
from sklearn.preprocessing import Normalizer
import multiprocessing

def run():
    path_rawdata = "F:\\TFG\\datasets\\raw_datasets\\datalake.csv"

    options = {
        "experiment": "own",
        "experiment_id": "prueba1",
        "data":{
            "aggregations": {
                "Points":"mean",
                "FTG":"mean","FTG_rival":"mean",
                "S":"mean", 
                "ST":"mean",
                "F":"mean", "C":"mean",
                "Y":"mean", "R":"mean",
                # "O":"mean", "HW":"mean", 
                "Side":"mean"
            },
            "features": [],
            "leagues": [],
            "lags": ["30D","120D","365D"],
            "min_samples": [2,10,20],
            "col_group": ["idTeam"],
            "col_window": "Date",
            # "split": 0.8,
            "sample": "sequential",
            "split_date":"2017-01-01",
            "save_data":False
        },
        "paths":{
            "data":"F:\\TFG\\datasets\\data_train\\data_"
        },
        "optimization":
        {
            "iterations":10,
        }
    }

    # IMPORT RAW DATA
    rawdata = ut.read_data(path_rawdata)
    options["data"]["scaler"] = Normalizer

    def objective(trial):
        model = training.MLP(trial,Dataset)
        val_loss = model.optimize_with_optuna(trial)
        return val_loss

    # CHOOSE EXPERIMENT
    if options["experiment"]=="own":
        Dataset = own.Dataflow_own(rawdata,options)
        print("Done")

        # Create a study and optimize the model
        study = optuna.create_study(direction='minimize')
        study.optimize(objective, n_trials=10)

        # Get the best hyperparameters
        best_params = study.best_trial.params
        print("Best Hyperparameters:", best_params)

        # Create the MLP model with the best hyperparameters
        best_model = training.MLP(study.best_trial)

multiprocessing.set_start_method('spawn', force=True)

# Spawn multiple processes for training or evaluation
process = multiprocessing.Process(target=run)
process.start()
process.join()
```
