Aquí guardamos en una lista con las temporadas de las cuales tenemos información en las diferentes ligas

```python
lista_temporadas = []

for i in range(93,99):
    t = str(i)+''+str(i+1)
    lista_temporadas = [t] + lista_temporadas
    
t = '9900'
lista_temporadas = [t] + lista_temporadas
    
for i in range(0,21):
        j = str(i+1)
        i = str(i)
        x = (2) if len(i) == 1 else (1)
        i = i.zfill(x)
        x = (2) if len(j) == 1 else (1)
        j = j.zfill(x)
        t = i+''+j
        lista_temporadas = [t] + lista_temporadas

print(lista_temporadas)
```

<h4>Empezamos a leer las distintas ligas que tenemos</h4>

He cambiado por codigo el nombre de los archivos para que indique liga y temporada a la que hace referencia (LIGA_temporada)

```python
# PRIMERA DIVISIÓN - la he hecho un poco diferente al resto (es la primera que hice jeje)

ligas['LaLiga'] = {}

for i in range(10,21):
    f = str(i+1)
    season = str(i) + f
    t = pd.read_csv(os.getcwd()+'\\SP1_'+ season +'.csv')
    ligas['LaLiga'][season] = t
    
for i in range(0,9):
    f = str(i+1)
    season = '0' + str(i) + '0' + f
    if season == '0405': # error en esa temporada
        continue
    t = pd.read_csv(os.getcwd()+'\\SP1_'+ season +'.csv')
    ligas['LaLiga'][season] = t
        
for i in range(93,99):
    f = str(i+1)
    season = str(i) + f
    t = pd.read_csv(os.getcwd()+'\\SP1_'+ season +'.csv')
    ligas['LaLiga'][season] = t
    
    
# excepciones 0910 y 9900
ligas['LaLiga']['0910'] = pd.read_csv(os.getcwd()+'\\SP1_'+ '0910' +'.csv') 
ligas['LaLiga']['9900'] = pd.read_csv(os.getcwd()+'\\SP1_'+ '9900' +'.csv')
```

Aquí he creado una leyenda a partir de un archivo .txt donde explica cada característica-

```python
# CREAMOS UNA LEYENDA DE TODOS LOS ATRIBUTOS
# Las leyendas cambian de una liga a otra (por las casas de apuestas)

leyenda = {}

f = open(os.getcwd() + '\\notesSpanishLeague.txt') # cogiendo los atributos de la liga española
lineas = f.readlines()

for col in ligas['LaLiga']['1920'].columns:  # cogiendo los atributos de la liga española
    for linea in lineas:
        if linea.find(col) == False:
            lim = linea.find('= ')
            lim = lim+2
            res = linea[lim:-1]
            leyenda[col] = res
            lineas.remove(linea)
```
